```python
import datetime

#Crea un script que te pregunta:
#Nombre
#Año de nacimiento
#Ahora con un cálculo
#Calcula la edad a partir del año de nacimiento :)
nombre = input("Cuál es tu nombre? ")
anio_nacimiento = int(input("Cuál es tu año de nacimiento? "))
anio_actual = datetime.datetime.now().year
correcto = False

while not correcto:
  if anio_nacimiento > anio_actual:
      print("Año de nacimiento incorrecto")
      anio_nacimiento = int(input("Cuál es tu año de nacimiento? "))
  else:
    edad = anio_actual - anio_nacimiento
    correcto = True


print(f"Hola {nombre} tu edad es {edad}")
```

```python
from datetime import datetime

from zoneinfo import ZoneInfo



# Hora de Nueva York

hora_ny = datetime.now(ZoneInfo("America/New_York"))

print("Nueva York:", hora_ny.strftime("%H:%M:%S"))



# Hora de Madrid

hora_madrid = datetime.now(ZoneInfo("Europe/Madrid"))

print("Madrid:", hora_madrid.strftime("%H:%M:%S"))



# Hora de Tokio

hora_tokio = datetime.now(ZoneInfo("Asia/Tokyo"))

print("Tokio:", hora_tokio.strftime("%H:%M:%S"))
```
