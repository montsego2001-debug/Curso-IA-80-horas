```python
# ============================================================
# 🐍 PYTHON LAB
# Librerías · Módulos · Clases · Métodos · Funciones
# ============================================================

import time
import ipywidgets as widgets

from IPython.display import display, clear_output
from datetime import datetime
from zoneinfo import ZoneInfo


# ============================================================
# 🏠 FUNCIÓN PRINCIPAL
# ============================================================

def inicio():

    clear_output(wait=True)

    # --------------------------------------------------------
    # TÍTULO
    # --------------------------------------------------------

    titulo = widgets.HTML("""
    <div style="
        background: linear-gradient(135deg, #1f4037, #99f2c8);
        color:white;
        padding:30px;
        border-radius:18px;
        text-align:center;
        font-family:Arial;
        margin-bottom:20px;
    ">

        <div style="font-size:42px;">🐍</div>

        <div style="
            font-size:32px;
            font-weight:bold;
        ">
            PYTHON LAB
        </div>

        <div style="
            font-size:18px;
            margin-top:8px;
        ">
            Entendiendo los conceptos básicos de Python
        </div>

    </div>
    """)

    # --------------------------------------------------------
    # EXPLICACIÓN
    # --------------------------------------------------------

    explicacion = widgets.HTML("""
    <div style="
        background:#f8f9fa;
        padding:20px;
        border-radius:15px;
        border:1px solid #ddd;
        font-family:Arial;
        margin-bottom:20px;
    ">

    <div style="
        font-size:22px;
        font-weight:bold;
        margin-bottom:15px;
    ">
        🔎 Vamos a aprender
    </div>

    <div style="
        font-size:18px;
        line-height:1.9;
    ">

    📚 <b>Librería</b> → conjunto de herramientas reutilizables.<br>

    📄 <b>Módulo</b> → archivo de Python que contiene código.<br>

    🏗️ <b>Clase</b> → sirve para crear objetos.<br>

    ⚙️ <b>Método</b> → función asociada a un objeto.<br>

    🔧 <b>Función</b> → realiza una tarea.

    </div>

    </div>
    """)

    # --------------------------------------------------------
    # BOTONES
    # --------------------------------------------------------

    btn_libreria = widgets.Button(
        description="📚 LIBRERÍA",
        layout=widgets.Layout(
            width="180px",
            height="55px"
        )
    )

    btn_modulo = widgets.Button(
        description="📄 MÓDULOS",
        layout=widgets.Layout(
            width="180px",
            height="55px"
        )
    )

    btn_clase = widgets.Button(
        description="🏗️ CLASE",
        layout=widgets.Layout(
            width="180px",
            height="55px"
        )
    )

    btn_metodo = widgets.Button(
        description="⚙️ MÉTODO",
        layout=widgets.Layout(
            width="180px",
            height="55px"
        )
    )

    btn_funcion = widgets.Button(
        description="🔧 FUNCIÓN",
        layout=widgets.Layout(
            width="180px",
            height="55px"
        )
    )

    botones = widgets.HBox(
        [
            btn_libreria,
            btn_modulo,
            btn_clase,
            btn_metodo,
            btn_funcion
        ],
        layout=widgets.Layout(
            justify_content="center",
            gap="10px"
        )
    )

    # --------------------------------------------------------
    # CONECTAMOS LOS BOTONES
    # --------------------------------------------------------

    btn_libreria.on_click(lambda b: ejemplo_libreria())
    btn_modulo.on_click(lambda b: ejemplo_modulos())
    btn_clase.on_click(lambda b: ejemplo_clase())
    btn_metodo.on_click(lambda b: ejemplo_metodo())
    btn_funcion.on_click(lambda b: ejemplo_funcion())

    display(
        titulo,
        explicacion,
        widgets.HTML("""
        <div style="
            text-align:center;
            font-size:20px;
            font-weight:bold;
            margin:20px;
        ">
        👇 Pulsa un concepto
        </div>
        """),
        botones
    )


# ============================================================
# 📚 1. LIBRERÍA
# ============================================================

def ejemplo_libreria():

    clear_output(wait=True)

    volver = widgets.Button(
        description="⬅️ Volver",
        layout=widgets.Layout(
            width="130px",
            height="40px"
        )
    )

    volver.on_click(lambda b: inicio())

    titulo = widgets.HTML("""
    <div style="
        background:#3498db;
        color:white;
        padding:22px;
        border-radius:15px;
        text-align:center;
        font-family:Arial;
    ">
        <div style="font-size:32px;">
            📚 LIBRERÍA
        </div>

        <div style="font-size:18px;">
            Un conjunto de herramientas reutilizables
        </div>
    </div>
    """)

    contenido = widgets.HTML("""
    <div style="
        padding:25px;
        font-family:Arial;
        font-size:18px;
        line-height:1.7;
    ">

    <h2>📚 Ejemplo: ipywidgets</h2>

    <p>
    <b>ipywidgets</b> es una librería que nos proporciona
    herramientas para crear elementos interactivos.
    </p>

    <div style="
        background:#272822;
        color:#f8f8f2;
        padding:20px;
        border-radius:10px;
        font-family:monospace;
        font-size:18px;
    ">
    import ipywidgets as widgets
    </div>

    <p>
    Con ella podemos crear botones, cuadros de texto,
    menús, etc.
    </p>

    <div style="
        background:#eaf2f8;
        padding:18px;
        border-radius:10px;
    ">

    📚 <b>ipywidgets</b><br>
    ↓<br>
    🏗️ Button<br>
    🏗️ Output<br>
    🏗️ HBox

    </div>

    </div>
    """)

    display(volver, titulo, contenido)


# ============================================================
# 📄 2. MÓDULOS
# ============================================================

def ejemplo_modulos():

    clear_output(wait=True)

    volver = widgets.Button(
        description="⬅️ Volver",
        layout=widgets.Layout(
            width="130px",
            height="40px"
        )
    )

    volver.on_click(lambda b: inicio())

    titulo = widgets.HTML("""
    <div style="
        background:#8e44ad;
        color:white;
        padding:22px;
        border-radius:15px;
        text-align:center;
        font-family:Arial;
    ">
        <div style="font-size:32px;">
            📄 MÓDULOS
        </div>

        <div style="font-size:18px;">
            Archivos que contienen código Python
        </div>
    </div>
    """)

    contenido = widgets.HTML("""
    <div style="
        padding:25px;
        font-family:Arial;
        font-size:18px;
        line-height:1.7;
    ">

    <h2>📄 Algunos módulos que hemos utilizado</h2>

    <div style="
        background:#f5eef8;
        padding:20px;
        border-radius:12px;
        margin-bottom:15px;
    ">

    📄 <b>time</b><br>
    └── 🔧 sleep()

    </div>

    <div style="
        background:#f5eef8;
        padding:20px;
        border-radius:12px;
        margin-bottom:15px;
    ">

    📄 <b>datetime</b><br>
    └── 🏗️ datetime

    </div>

    <div style="
        background:#f5eef8;
        padding:20px;
        border-radius:12px;
        margin-bottom:15px;
    ">

    📄 <b>zoneinfo</b><br>
    └── 🏗️ ZoneInfo

    </div>

    <p>
    Por ejemplo:
    </p>

    <div style="
        background:#272822;
        color:#f8f8f2;
        padding:20px;
        border-radius:10px;
        font-family:monospace;
        font-size:18px;
    ">

    import time<br><br>

    from datetime import datetime<br><br>

    from zoneinfo import ZoneInfo

    </div>

    </div>
    """)

    display(volver, titulo, contenido)


# ============================================================
# 🏗️ 3. CLASE
# ============================================================

def ejemplo_clase():

    clear_output(wait=True)

    volver = widgets.Button(
        description="⬅️ Volver",
        layout=widgets.Layout(
            width="130px",
            height="40px"
        )
    )

    volver.on_click(lambda b: inicio())

    titulo = widgets.HTML("""
    <div style="
        background:#e67e22;
        color:white;
        padding:22px;
        border-radius:15px;
        text-align:center;
        font-family:Arial;
    ">
        <div style="font-size:32px;">
            🏗️ CLASE
        </div>

        <div style="font-size:18px;">
            Una clase sirve para crear objetos
        </div>
    </div>
    """)

    contenido = widgets.HTML("""
    <div style="
        padding:25px;
        font-family:Arial;
        font-size:18px;
        line-height:1.7;
    ">

    <h2>🏗️ Ejemplo: Button</h2>

    <div style="
        background:#272822;
        color:#f8f8f2;
        padding:20px;
        border-radius:10px;
        font-family:monospace;
        font-size:18px;
    ">

    btn = widgets.Button(<br>
    &nbsp;&nbsp;description="Púlsame"<br>
    )

    </div>

    <p>
    🏗️ <b>Button</b> es una clase.
    </p>

    <p>
    La clase nos permite crear un botón.
    </p>

    <p>
    El botón concreto que creamos será un:
    </p>

    <div style="
        background:#eaf2f8;
        padding:20px;
        border-radius:12px;
        text-align:center;
        font-size:22px;
    ">

    🧱 <b>objeto</b>

    </div>

    </div>
    """)

    boton = widgets.Button(
        description="🖱️ Púlsame",
        button_style="success",
        layout=widgets.Layout(
            width="180px",
            height="50px"
        )
    )

    display(volver, titulo, contenido, boton)


# ============================================================
# ⚙️ 4. MÉTODO
# ============================================================

def ejemplo_metodo():

    clear_output(wait=True)

    volver = widgets.Button(
        description="⬅️ Volver",
        layout=widgets.Layout(
            width="130px",
            height="40px"
        )
    )

    volver.on_click(lambda b: inicio())

    titulo = widgets.HTML("""
    <div style="
        background:#16a085;
        color:white;
        padding:22px;
        border-radius:15px;
        text-align:center;
        font-family:Arial;
    ">
        <div style="font-size:32px;">
            ⚙️ MÉTODO
        </div>

        <div style="font-size:18px;">
            Una función asociada a un objeto
        </div>
    </div>
    """)

    boton = widgets.Button(
        description="🖱️ Púlsame",
        button_style="success",
        layout=widgets.Layout(
            width="180px",
            height="50px"
        )
    )

    resultado = widgets.Output()

    def cuando_pulse(b):

        with resultado:

            clear_output()

            print("🎉 ¡Has pulsado el botón!")
            print()
            print("🧱 boton → objeto")
            print("⚙️ on_click() → método")
            print("🔧 cuando_pulse() → función")

    boton.on_click(cuando_pulse)

    contenido = widgets.HTML("""
    <div style="
        padding:25px;
        font-family:Arial;
        font-size:18px;
        line-height:1.7;
    ">

    <h2>⚙️ Ejemplo</h2>

    <div style="
        background:#272822;
        color:#f8f8f2;
        padding:20px;
        border-radius:10px;
        font-family:monospace;
        font-size:18px;
    ">

    boton.on_click(cuando_pulse)

    </div>

    <p>
    🧱 <b>boton</b> → objeto
    </p>

    <p>
    ⚙️ <b>on_click()</b> → método del objeto
    </p>

    <p>
    🔧 <b>cuando_pulse()</b> → función que se ejecutará
    </p>

    <p>
    👇 Pulsa el botón para comprobarlo.
    </p>

    </div>
    """)

    display(
        volver,
        titulo,
        contenido,
        boton,
        resultado
    )


# ============================================================
# 🔧 5. FUNCIÓN
# ============================================================

def ejemplo_funcion():

    clear_output(wait=True)

    volver = widgets.Button(
        description="⬅️ Volver",
        layout=widgets.Layout(
            width="130px",
            height="40px"
        )
    )

    volver.on_click(lambda b: inicio())

    titulo = widgets.HTML("""
    <div style="
        background:#c0392b;
        color:white;
        padding:22px;
        border-radius:15px;
        text-align:center;
        font-family:Arial;
    ">
        <div style="font-size:32px;">
            🔧 FUNCIÓN
        </div>

        <div style="font-size:18px;">
            Una función realiza una tarea
        </div>
    </div>
    """)

    def saludar(nombre):
        return "Hola " + nombre + " 👋"

    resultado = saludar("Montse")

    contenido = widgets.HTML(f"""
    <div style="
        padding:25px;
        font-family:Arial;
        font-size:18px;
        line-height:1.7;
    ">

    <h2>🔧 Ejemplo: nuestra propia función</h2>

    <div style="
        background:#272822;
        color:#f8f8f2;
        padding:20px;
        border-radius:10px;
        font-family:monospace;
        font-size:18px;
    ">

    def saludar(nombre):<br>
    &nbsp;&nbsp;&nbsp;&nbsp;return "Hola " + nombre<br><br>

    saludar("Montse")

    </div>

    <p>
    🔧 <b>saludar()</b> → función
    </p>

    <p>
    La función recibe un nombre y devuelve un saludo.
    </p>

    <div style="
        background:#fdecea;
        padding:20px;
        border-radius:12px;
        font-size:22px;
        text-align:center;
    ">

    🎯 Resultado: <b>{resultado}</b>

    </div>

    </div>
    """)

    display(volver, titulo, contenido)


# ============================================================
# 🚀 ARRANCAR
# ============================================================

inicio()
```
