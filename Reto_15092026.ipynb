````python
#VERSION PARA HACERLO BONITO
#vamos a crear funciones de EF, fuerza, velocidad, flexibilidad, coordinación, nombre alumno, edad
#librería para hacer botones interactivos
import ipywidgets as widgets
#librería para pintarlos
from IPython.display import display, clear_output

alumnos = []
def crear_alumnos():
  for x in range(1, 18):
      nombre = input("Nombre: ")
      while True:
          try:
              edad = int(input("Edad: "))
              break
          except ValueError:
              print("Error: la edad debe ser un número")
      while True:
          try:
              repeticiones = int(input("Repeticiones: "))
              break
          except ValueError:
              print("Error: la edad debe ser un número")
      ejercicio = input("Ejercicio: ")
      alumno = {
          "nombre": nombre,
          "edad": edad,
          "repeticiones": repeticiones,
          "ejercicio": ejercicio
      }
      alumnos.append(alumno)
      seguimos = input("¿Quieres seguir? (s/n): ")
      if seguimos == "n":
          break

crear_alumnos()

#ahora vamos a crear tantos botones como alumnos haya
botones = []
for alumno in alumnos:
    boton = widgets.Button(
        description=alumno["nombre"],
        layout=widgets.Layout(
            width="180px",
            height="55px",
            background="#fff",
            color="#ffff00"
        ),
        button_style="info"
    )
    botones.append(boton)

display(widgets.HBox(botones))
# para pintar toda la lista y su estructura print(alumnos)
#for alumno in alumnos:
#    print(alumno["nombre"])
#    print(alumno["edad"])
#    print(alumno["repeticiones"])
#    print(alumno["ejercicio"])
#
````
